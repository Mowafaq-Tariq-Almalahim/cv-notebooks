```python
import cv2 as cv
import numpy as np
```

```python
img = cv.imread("ziggs.png")
```

```python
img = cv.resize(img, (960,480), interpolation=cv.INTER_AREA)
```

```python
cv.imshow("image", img)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
height, width = img.shape[:2]


# 100 pixels to the right, 50 pixels down
T = np.float32([
    [1, 0, 100],
    [0, 1, 50] ]
) 
```

```python
translated = cv.warpAffine(img, T, (width, height))
```

```python
cv.imshow("the Transofmation result", translated)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
center = (width//2, height//2)
R = cv.getRotationMatrix2D(center, 45, 1.0) # the point we want to rotate around and the degree we want to rotate the center around and the scale factor
```

```python
rotated = cv.warpAffine(img, R, (width, height))
```

```python
cv.imshow("rotated image", rotated)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
cropped = img[100:300, 200:500]
```

```python
cv.imshow("cropped image", cropped)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
flipped = cv.flip(img, 0)
```

```python
cv.imshow("flipped image", flipped)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
small = cv.resize(img, (480,240))
```

```python
cv.imshow("small image", small)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
brightened = cv.add(img, 50)
```

```python
cv.imshow("Original", img)
cv.imshow("Brightened", brightened)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
darker = cv.subtract(img, 50)
```

```python
cv.imshow("Original", img)
cv.imshow("Darker", darker)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
blended = cv.addWeighted(img, 0.1, darker, 0.9, 0)
```

```python
cv.imshow("blended", blended)
cv.waitKey(0)
cv.destroyAllWindows()
```
