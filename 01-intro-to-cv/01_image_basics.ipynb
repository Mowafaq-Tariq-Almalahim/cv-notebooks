```python
import numpy as np
import cv2 as cv
```

```python
img = cv.imread("ziggs.png")
```

```python
img = cv.resize(img, (960,480), interpolation=cv.INTER_AREA)
```

```python
cv.imshow("resized image", img)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
img.shape
```

```python
resized = img.copy()
```

```python
gray = cv.cvtColor(resized, cv.COLOR_BGR2GRAY)
```

```python
cv.imshow("gray image",gray)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
cv.imwrite("gray_ziggs.png", gray)
```
