```python
import cv2 as cv
import numpy as np
```

```python
img = cv.imread("ziggs.png")
```

```python
img = cv.resize(img, (960,480), interpolation=cv.INTER_AREA)
```

```python
inverted = cv.bitwise_not(img)
```

```python
cv.imshow("inverted", inverted)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# create a new image for Compassion

other = np.zeros_like(img)
cv.rectangle( other, (100,100), (400,300), (255,255,255), -1)
```

```python
and_img = cv.bitwise_and(img, other)
or_img = cv.bitwise_or(img, other)
xor_img = cv.bitwise_xor(img, other)
```

```python
cv.imshow("Original", img)
cv.imshow("AND", and_img)
cv.imshow("OR", or_img)
cv.imshow("XOR", xor_img)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
height, width = img.shape[:2]
```

```python
mask = np.zeros((height, width), dtype=np.uint8)
cv.circle(mask, (width//2, height//2), 100, 1, -1)
```

```python
cv.imshow("masked image", mask)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
masked = cv.bitwise_and(img, img, mask=mask)
```

```python
cv.imshow("og", img)
cv.imshow("masked", masked)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
mask = np.zeros((height, width), dtype=np.uint8)
cv.rectangle(mask, (200,100), (700,400), 255, -1)
masked = cv.bitwise_and(img, img, mask=mask)
mask1 = np.zeros((height, width), dtype=np.uint8)
cv.circle(mask1, (width//2, height//2), 250, 1, -1)
masked1 = cv.bitwise_and(img, img, mask=mask1)
fnl = cv.bitwise_or(masked, masked1)
# cv.imshow("og", masked1)
# cv.imshow("masked", masked)
cv.imshow("fnl",fnl)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
mask = np.zeros((height, width), dtype=np.uint8)
cv.rectangle(mask, (200,100), (700,400), 255, -1)
mask1 = np.zeros((height, width), dtype=np.uint8)
cv.circle(mask1, (width//2, height//2), 250, 255, -1)
```

```python
both = cv.bitwise_and(mask, mask1)
either = cv.bitwise_or(mask, mask1)
exclusive = cv.bitwise_xor(mask, mask1)

cv.imshow("Intersection", both)
cv.imshow("Union", either)
cv.imshow("Exclusive regions", exclusive)

cv.waitKey(0)
cv.destroyAllWindows()
```
